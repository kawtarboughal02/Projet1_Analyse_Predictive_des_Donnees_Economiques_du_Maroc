```python
import pandas as pd
import numpy as np
```

```python
inflation = pd.read_csv('../data/raw/inflation.csv', skiprows=4)
chomage = pd.read_csv('../data/raw/chomage.csv', skiprows=4)
pib = pd.read_csv('../data/raw/pib.csv', skiprows=4)
```

```python
def nettoyer_indicateur(df, nom_colonne, annee_debut=2000, annee_fin=2024):
    
    colonnes_annees = [str(a) for a in range(annee_debut, annee_fin + 1)]
    colonnes_annees = [c for c in colonnes_annees if c in df.columns]

    df_maroc = df[df['Country Name'] == 'Morocco'][colonnes_annees]

    serie = df_maroc.T
    serie.columns = [nom_colonne]
    serie.index = serie.index.astype(int)
    serie = serie.sort_index()
    serie[nom_colonne] = pd.to_numeric(serie[nom_colonne], errors='coerce')

    return serie
```

```python
inflation_t = nettoyer_indicateur(inflation, 'inflation_pct')
chomage_t = nettoyer_indicateur(chomage, 'chomage_pct')
pib_t = nettoyer_indicateur(pib, 'pib')

inflation_t.head()
```

```python
df = pd.concat([pib_t, inflation_t, chomage_t], axis=1)
df.head(10)
```

```python
print("Valeurs manquantes par colonne :")
print(df.isnull().sum())
```

```python
df = df.interpolate(method='linear')
print("Valeurs manquantes après traitement :")
print(df.isnull().sum())
```

```python
df.to_csv('../data/processed/donnees_economiques_maroc.csv')
print("Fichier sauvegardé avec succès.")
df.tail()
```
